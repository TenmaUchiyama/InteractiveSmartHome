```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Evaluator (for correlation / causal analysis)

- VOICE_LOG から participant×condition×task データを生成
- arrange_color_map.json があれば色数を付与
- タスク単位と参加者×条件単位のCSVを出力
"""

import os, re, json, glob, math
import pandas as pd
import numpy as np
from pathlib import Path

# ====== パス設定 ======
VOICE_ROOT = "./VOICE_LOG"
ARRANGE_COLOR_MAP_PATH = "./arrange_color_map.json"
OUT_DIR = "./imwut_results"
os.makedirs(OUT_DIR, exist_ok=True)

# ====== 条件マップ ======
COND_MAP = {1: "SR", 2: "P+SR", 3: "Pointing", 4: "Label"}
COND_ORDER = ["SR", "P+SR", "Pointing", "Label"]

# ====== ユーティリティ ======
def safe_float(x, default=None):
    try:
        if x in (None, "", "NaN"): return default
        return float(x)
    except Exception:
        return default

def dedupe_attempts(attempts):
    """attemptId + command + outputDevices が同一の連続行を1回に圧縮"""
    out, prev = [], None
    for a in attempts or []:
        aid, cmd = a.get("attemptId"), a.get("userCommand")
        outs = tuple(a.get("outputDevices") or [])
        key = (aid, cmd, outs)
        if key != prev:
            out.append(a)
            prev = key
    return out

def parse_condition_from_filename(name: str):
    m = re.match(r"^\s*([1-4])(?:[\-_].*)?\.json$", name)
    return int(m.group(1)) if m else None

def parse_participant_from_path(path: Path):
    """P1_1.json → participant=1"""
    m = re.search(r"P(\d+)", str(path))
    return int(m.group(1)) if m else None

# ====== arrange_color_map 読み込み ======
arrange_color_map = {}
if os.path.exists(ARRANGE_COLOR_MAP_PATH):
    with open(ARRANGE_COLOR_MAP_PATH, "r", encoding="utf-8") as f:
        arrange_color_map = json.load(f)

def get_color_count(taskId):
    """タスクIDに対応する色の種類数を返す（辞書型でもリスト型でもOK）"""
    if not isinstance(arrange_color_map, dict):
        return np.nan
    entry = arrange_color_map.get(taskId)
    if entry is None:
        return np.nan
    # 辞書型 {"colors": [...]} or リスト型 [...] の両方を許容
    if isinstance(entry, dict):
        colors = entry.get("colors", [])
    elif isinstance(entry, list):
        colors = entry
    else:
        return np.nan
    if isinstance(colors, list):
        return len(set(colors))
    return np.nan


# ====== メイン ======
records = []

for json_path in sorted(Path(VOICE_ROOT).rglob("*.json")):
    if json_path.name.endswith(".meta.json"):
        continue
    cond_num = parse_condition_from_filename(json_path.name)
    if cond_num is None:
        continue
    condition = COND_MAP.get(cond_num, str(cond_num))
    participant = parse_participant_from_path(json_path)

    with open(json_path, "r", encoding="utf-8") as f:
        try:
            tasks = json.load(f)
        except Exception:
            continue

    for task in tasks:
        taskId = task.get("taskId")
        total_time = safe_float(task.get("totalElapsedTime"), np.nan)
        attempts = dedupe_attempts(task.get("taskAttempts"))
        n_cmds = len(attempts)
        n_actuations = sum(len(a.get("outputDevices", [])) for a in attempts)
        n_devices_final = len(attempts[-1].get("outputDevices", [])) if attempts else 0
        n_colors_required = get_color_count(taskId)

        avg_cmd_time = total_time / n_cmds if n_cmds > 0 else np.nan
        logT = math.log(total_time) if total_time and total_time > 0 else np.nan

        records.append({
            "participant": participant,
            "condition": condition,
            "taskId": taskId,
            "T_task_sec": total_time,
            "logT_task": logT,
            "N_cmds": n_cmds,
            "N_actuations": n_actuations,
            "n_devices_final": n_devices_final,
            "n_colors_required": n_colors_required,
            "avg_cmd_time": avg_cmd_time
        })

# ====== タスク単位で保存 ======
df = pd.DataFrame(records)
csv_task = os.path.join(OUT_DIR, "task_summary_enriched.csv")
df.to_csv(csv_task, index=False, encoding="utf-8-sig")

# ====== 参加者×条件レベルで集約 ======
group_cols = ["participant", "condition"]
agg_df = (
    df.groupby(group_cols)
    .agg({
        "T_task_sec": "median",
        "logT_task": "median",
        "N_cmds": "median",
        "N_actuations": "median",
        "n_devices_final": "median",
        "n_colors_required": "median",
        "avg_cmd_time": "median"
    })
    .reset_index()
)
csv_pc = os.path.join(OUT_DIR, "task_summary_pc.csv")
agg_df.to_csv(csv_pc, index=False, encoding="utf-8-sig")

print("✅ Exported:")
print(" -", csv_task)
print(" -", csv_pc)
```

```python
# --- 論文本文用 Numerical Analysis プロット（完全版：個別ブラケット縦オフセット対応） ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
import scikit_posthocs as sp
from scipy.stats import friedmanchisquare

# =============================================================================
# 0) 基本設定（★ここを変える：パス・並び順・背景・凡例）
# =============================================================================
CSV_PATH   = "./imwut_results/participant_condition_summary.csv"  # ★CSVパス
COND_ORDER = ["SR", "P+SR", "Pointing", "Label"]                  # ★条件の並び順
BG_COLOR   = "#f7f7f7"                                            # 背景色
PALETTE    = [cm.get_cmap("Set2")(i) for i in range(len(COND_ORDER))]  # 色

# =============================================================================
# 1) Friedman + Nemenyi（変えない）
# =============================================================================
def friedman_nemenyi(df, metric):
    pivot = df.pivot(index="participant", columns="condition", values=metric)
    pivot = pivot.reindex(columns=COND_ORDER).dropna()
    stat, p = friedmanchisquare(*[pivot[c].values for c in COND_ORDER])
    posthoc = sp.posthoc_nemenyi_friedman(pivot)
    pairs_by_alpha = {"0.01": [], "0.05": []}
    for i, a in enumerate(COND_ORDER):
        for j, b in enumerate(COND_ORDER):
            if i >= j:
                continue
            pval = float(posthoc.loc[a, b])
            if pval < 0.01:
                pairs_by_alpha["0.01"].append((i + 1, j + 1))
            elif pval < 0.05:
                pairs_by_alpha["0.05"].append((i + 1, j + 1))
    return stat, p, pairs_by_alpha, pivot

# =============================================================================
# 2) 有意差ブラケット（★ここが本体：個別オフセットOK）
# =============================================================================
def add_sig_brackets(
    ax,
    pairs_by_alpha,
    id_to_pos=None,
    star_map=None,                 # {"0.01":"**","0.05":"*"}
    top_pad_ratio=0.06,            # 図上端→最初のブラケットまで（比率）
    step_ratio=0.06,               # 段間（比率）
    tick_height_ratio=0.02,        # ブラケット縦棒の高さ（比率）

    # <<<==============================================================>>>
    # ★★★ スター(＊,＊＊)の縦位置オフセット ★★★
    # ・単位は y軸レンジに対する「比率」
    # ・描画位置: y_star = y_bracket_top + tick_h + (yr * text_offset_ratio)
    # ・負の値 → 星が線に近づく/重なる、正の値 → 星が上へ離れる
    # ・例: -0.01 (重ねる), 0.012 (少し離す)
    # <<<==============================================================>>>
    text_offset_ratio=0.0,

    lw=1.2,
    star_fontsize=10,
    title_pad=4,
    alpha_priority=("0.05", "0.01"),
    allow_touch=True,
    pack_by="length",
    touch_gap=0.06,                # 同レベルで接する端点を左右に分ける量（x座標単位）
    inner_pad=0.02,                # すべてのブラケット両端に入れる微小インセット

    # -----------------------------------------------------------------
    # ★ここを変える：各ブラケットの縦位置“個別”オフセット
    #   1) ペアごと: {(1,3): 0.02, (2,4): 0.01} など（比率）
    #   2) レベルごと: {0: 0.00, 1: 0.01, 2: 0.02} など（比率）
    #   3) 条件付き関数: lambda akey,pair,lv,x1,x2: 0.01 if pair==(1,3) else 0.0
    # いずれも“上方向が正”。start に重なるブラケットだけ上に逃がせます。
    # -----------------------------------------------------------------
    per_pair_offset_ratio=None,
    per_level_offset_ratio=None,
    offset_fn=None,
):
    import numpy as np
    from collections import defaultdict

    if star_map is None:
        star_map = {"0.01": "**", "0.05": "*"}

    # カテゴリID→x座標
    if id_to_pos is None:
        all_ids = []
        for v in pairs_by_alpha.values():
            all_ids.extend([a for pair in v for a in pair])
        n = max(all_ids) if all_ids else 0
        id_to_pos = {i: i for i in range(1, n + 1)}

    # 全ペアを正規化して集約
    all_pairs = []
    for alpha_key, vec in pairs_by_alpha.items():
        for (a, b) in vec:
            x1, x2 = id_to_pos[a], id_to_pos[b]
            if x1 > x2:
                x1, x2 = x2, x1
            all_pairs.append((alpha_key, (a, b), x1, x2))
    if not all_pairs:
        ax.set_title(ax.get_title(), pad=title_pad)
        return

    # alpha優先で詰める
    alpha_rank = {k: i for i, k in enumerate(alpha_priority)}
    def alpha_key_order(k): return alpha_rank.get(k, len(alpha_priority))

    assigned = []
    level_offset = 0
    for akey in sorted(set(k for k, *_ in all_pairs), key=alpha_key_order):
        group = [(k, pair, x1, x2) for (k, pair, x1, x2) in all_pairs if k == akey]
        if pack_by == "length":
            group.sort(key=lambda t: (t[2], (t[3]-t[2])))  # x1昇順, 長さ昇順
        else:
            group.sort(key=lambda t: (t[2], t[3]))         # x1昇順, x2昇順

        # --- レベル割当（区間スケジューリング）---
        end_per_level = []   # 各レベルの右端x
        local_assigned = []
        for (k, pair, x1, x2) in group:
            placed = False
            for lv, end_x in enumerate(end_per_level):
                ok = (x1 >= end_x) if allow_touch else (x1 > end_x)
                if ok:
                    end_per_level[lv] = x2
                    local_assigned.append((k, pair, x1, x2, lv))
                    placed = True
                    break
            if not placed:
                end_per_level.append(x2)
                local_assigned.append((k, pair, x1, x2, len(end_per_level)-1))

        for (k, pair, x1, x2, lv) in local_assigned:
            assigned.append((k, pair, x1, x2, level_offset + lv))
        level_offset += len(end_per_level)

    n_levels = (max(lv for *_, lv in assigned) + 1) if assigned else 0

    # ---- y軸の余白調整（個別オフセット分も考慮して切れないように）----
    ymin, ymax = ax.get_ylim()
    yr = ymax - ymin

    def _pair_key(pair):
        a, b = pair
        return (a, b) if a < b else (b, a)

    def _get_offset_ratio(akey, pair, lv, x1, x2):
        r = 0.0
        if per_level_offset_ratio and lv in per_level_offset_ratio:
            r += float(per_level_offset_ratio[lv])
        if per_pair_offset_ratio:
            r += float(per_pair_offset_ratio.get(_pair_key(pair), 0.0))
        if offset_fn:
            r += float(offset_fn(akey, pair, lv, x1, x2) or 0.0)
        return r

    per_bracket_offsets = [
        _get_offset_ratio(akey, pair, lv, x1, x2)
        for (akey, pair, x1, x2, lv) in assigned
    ]
    max_pos_off = max([o for o in per_bracket_offsets if o > 0.0], default=0.0)

    needed_extra = yr * (
        top_pad_ratio
        + max(0, n_levels-1)*step_ratio
        + tick_height_ratio
        + max(0.0, text_offset_ratio)
        + max_pos_off         # ★個別上乗せ分（正側）も余白に確保
        + 0.004
    )
    ax.set_ylim(ymin, ymax + needed_extra)

    # ---- y座標の基準を再計算 ----
    ymin, ymax = ax.get_ylim()
    yr = ymax - ymin
    base_y  = ymax - yr * (
        top_pad_ratio + max(0, n_levels-1)*step_ratio
        + tick_height_ratio + max(0.0, text_offset_ratio)
    )
    tick_h  = yr * tick_height_ratio
    text_dy = yr * text_offset_ratio

    # ---- 同レベル内の“接触”微調整（横方向）----
    by_level = defaultdict(list)
    for idx, item in enumerate(assigned):
        by_level[item[4]].append(idx)

    left_pad  = [inner_pad] * len(assigned)
    right_pad = [inner_pad] * len(assigned)
    tol = 1e-9
    for lv, idxs in by_level.items():
        idxs_sorted = sorted(idxs, key=lambda i: (assigned[i][2], assigned[i][3]))
        for i_prev, i_next in zip(idxs_sorted[:-1], idxs_sorted[1:]):
            _, _, x1_prev, x2_prev, _ = assigned[i_prev]
            _, _, x1_next, x2_next, _ = assigned[i_next]
            if abs(x2_prev - x1_next) <= tol:
                right_pad[i_prev] += touch_gap/2.0
                left_pad[i_next]  += touch_gap/2.0

    # ---- 描画（下段→上段、同段は左→右）----
    draw_list = sorted(
        [(i, *assigned[i]) for i in range(len(assigned))],
        key=lambda t: (t[5], t[3], t[4]-t[3])   # level, x1, length
    )

    for row in draw_list:
        idx, akey, pair, x1, x2, lv = row
        xp1 = x1 + left_pad[idx]
        xp2 = x2 - right_pad[idx]
        if xp2 - xp1 < 0.02:
            mid = (x1 + x2) / 2.0
            xp1, xp2 = mid - 0.01, mid + 0.01

        # ★各ブラケット固有の縦オフセット（比率→実寸）を加える
        y_offset = per_bracket_offsets[idx] * yr

        y = base_y + lv * step_ratio * yr + y_offset
        ax.plot([xp1, xp1, xp2, xp2],
                [y,   y+tick_h, y+tick_h, y],
                color="black", linewidth=lw)
        ax.text((xp1+xp2)/2.0, y + tick_h + text_dy,
                star_map.get(akey, "*"),
                ha="center", va="bottom", fontsize=star_fontsize)

    ax.set_title(ax.get_title(), pad=title_pad)

# =============================================================================
# 3) 描画（★ここを変える：個別ブラケットを上に逃がす指定はこの呼び出しで）
# =============================================================================
plt.rcParams.update({
    "figure.facecolor": BG_COLOR,
    "axes.facecolor": BG_COLOR,
    "font.size": 13,
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
})

df = pd.read_csv(CSV_PATH)
print(f"Loaded: {len(df)} rows")

fig, axes = plt.subplots(3, 1, figsize=(6.5, 11))

# ---------------------- 1) Time ----------------------
metric = "T_task_sec"
stat, p, pairs, pivot = friedman_nemenyi(df, metric)
vals = [pivot[c].dropna().values for c in COND_ORDER]
bp = axes[0].boxplot(vals, labels=COND_ORDER, patch_artist=True, showfliers=False)
for i, patch in enumerate(bp['boxes']):
    patch.set_facecolor(PALETTE[i]); patch.set_alpha(0.85)
    patch.set_edgecolor("black");    patch.set_linewidth(1.2)
for w in bp['whiskers']: w.set_color("black")
for c in bp['caps']:     c.set_color("black")
for m in bp['medians']:  m.set_color("black")
for i, v in enumerate(vals, 1):
    xs = np.random.normal(loc=i, scale=0.04, size=len(v))
    axes[0].plot(xs, v, 'o', color='black', alpha=0.75, ms=4)

axes[0].set_title(f"Task Completion Time ($T_{{task}}$) — χ²={stat:.2f}, p={p:.3g}")
axes[0].set_ylabel("Time [s]")
axes[0].grid(alpha=0.35, linestyle='--')

# ★ここを変える：Time 図で「startに重なるブラケットだけ」上に逃がしたいとき
#   例: (1,3) と (2,4) のブラケットをそれぞれ 1.5% / 1.0% 上へ
time_pair_offset = {(1,3): 0.015, (2,4): 0.010}  # ←不要なら {} に
add_sig_brackets(
    axes[0], pairs,
    text_offset_ratio=0.0,           # 星の上下（比率）。線から離したいなら 0.012 など
    top_pad_ratio=0.06,              # ブラケット群の開始高さ（まとめて上げる）
    step_ratio=0.06,                 # 段間
    tick_height_ratio=0.02,          # 縦棒高さ
    per_pair_offset_ratio=time_pair_offset,  # ★各ブラケット固有の縦オフセット（比率）
    # per_level_offset_ratio={1:0.01},      # 段ごとにまとめて上げたい場合はこっち
    # offset_fn=lambda akey,pair,lv,x1,x2: 0.01 if (x2-x1)<1.1 else 0.0,
)

# ---------------------- 2) N_cmds ----------------------
metric = "N_cmds"
stat, p, pairs, pivot = friedman_nemenyi(df, metric)
vals = [pivot[c].dropna().values for c in COND_ORDER]
bp = axes[1].boxplot(vals, labels=COND_ORDER, patch_artist=True, showfliers=False)
for i, patch in enumerate(bp['boxes']):
    patch.set_facecolor(PALETTE[i]); patch.set_alpha(0.85)
    patch.set_edgecolor("black");    patch.set_linewidth(1.2)
for w in bp['whiskers']: w.set_color("black")
for c in bp['caps']:     c.set_color("black")
for m in bp['medians']:  m.set_color("black")
for i, v in enumerate(vals, 1):
    xs = np.random.normal(loc=i, scale=0.04, size=len(v))
    axes[1].plot(xs, v, 'o', color='black', alpha=0.75, ms=4)

axes[1].set_title(f"Number of Commands ($N_{{cmds}}$) — χ²={stat:.2f}, p={p:.3g}")
axes[1].set_ylabel("Count")
axes[1].grid(alpha=0.35, linestyle='--')

# ★ここを変える：N_cmds 図の個別オフセット（必要なら）
add_sig_brackets(
    axes[1], pairs,
    text_offset_ratio=0.0,
    top_pad_ratio=0.06,
    step_ratio=0.06,
    tick_height_ratio=0.02,
    per_pair_offset_ratio={},        # 例: {(1,4):0.02}
)

# ---------------------- 3) 散布図 ----------------------
pivot_T = df.pivot(index="participant", columns="condition",
                   values="T_task_sec").reindex(columns=COND_ORDER)
pivot_N = df.pivot(index="participant", columns="condition",
                   values="N_cmds").reindex(columns=COND_ORDER)
x, y, c = [], [], []
for i, cond in enumerate(COND_ORDER):
    merged = pd.concat([pivot_T[cond].rename("T"),
                        pivot_N[cond].rename("N")], axis=1).dropna()
    x.extend(merged["N"]); y.extend(merged["T"]); c.extend([PALETTE[i]] * len(merged))
axes[2].scatter(x, y, color=c, edgecolor="black", alpha=0.8, s=40)
axes[2].set_xlabel("Number of Commands ($N_{cmds}$)")
axes[2].set_ylabel("Task Completion Time ($T_{task}$)")
axes[2].set_title("Relationship between $N_{cmds}$ and $T_{task}$")
axes[2].grid(alpha=0.35, linestyle='--')

plt.tight_layout()
plt.show()
```

```python
# --- 論文本文用 Numerical Analysis プロット（デザイン統一 & 完全版） ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
import scikit_posthocs as sp
from scipy.stats import friedmanchisquare

# ===== パス設定 / 定数 =====
CSV_PATH = "./imwut_results/participant_condition_summary.csv"
COND_ORDER = ["SR", "P+SR", "Pointing", "Label"]
BG_COLOR = "#f7f7f7"        # 背景色
HEADROOM_RATIO_TIME = 0.12  # ← Time図だけ上側に追加する余白（比率）。不要なら 0 に

# ===== データ読み込み =====
df = pd.read_csv(CSV_PATH)
print(f"Loaded: {len(df)} rows")

# ===== Friedman + Nemenyi =====
def friedman_nemenyi(df, metric):
    pivot = df.pivot(index="participant", columns="condition", values=metric)
    pivot = pivot.reindex(columns=COND_ORDER).dropna()
    stat, p = friedmanchisquare(*[pivot[c].values for c in COND_ORDER])
    posthoc = sp.posthoc_nemenyi_friedman(pivot)
    pairs_by_alpha = {"0.01": [], "0.05": []}
    for i, a in enumerate(COND_ORDER):
        for j, b in enumerate(COND_ORDER):
            if i >= j:
                continue
            pval = float(posthoc.loc[a, b])
            if pval < 0.01:
                pairs_by_alpha["0.01"].append((i + 1, j + 1))
            elif pval < 0.05:
                pairs_by_alpha["0.05"].append((i + 1, j + 1))
    return stat, p, pairs_by_alpha, pivot



# ===== 背景設定 / 体裁 =====
plt.rcParams.update({
    "figure.facecolor": BG_COLOR,
    "axes.facecolor": BG_COLOR,
    "font.size": 13,
    "axes.titlesize": 15,
    "axes.labelsize": 13,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
})

# ===== パレット =====
palette = [cm.get_cmap("Set2")(i) for i in range(len(COND_ORDER))]

# ===== 図作成（縦3段） =====
fig, axes = plt.subplots(3, 1, figsize=(6.5, 11))

# === 1️⃣ T_task_sec ===
metric = "T_task_sec"
stat, p, pairs, pivot = friedman_nemenyi(df, metric)
vals = [pivot[c].dropna().values for c in COND_ORDER]
bp = axes[0].boxplot(vals, labels=COND_ORDER, patch_artist=True, showfliers=False)
for i, patch in enumerate(bp['boxes']):
    patch.set_facecolor(palette[i]); patch.set_alpha(0.85)
    patch.set_edgecolor("black");    patch.set_linewidth(1.2)
for whisker in bp['whiskers']: whisker.set_color("black")
for cap in bp['caps']:         cap.set_color("black")
for median in bp['medians']:   median.set_color("black")
for i, v in enumerate(vals, 1):
    xs = np.random.normal(loc=i, scale=0.04, size=len(v))
    axes[0].plot(xs, v, 'o', color='black', alpha=0.75, ms=4)
axes[0].set_title(f"Task Completion Time ($T_{{task}}$) — χ²={stat:.2f}, p={p:.3g}")
axes[0].set_ylabel("Time [s]")
axes[0].grid(alpha=0.35, linestyle='--')
add_sig_brackets(axes[0], pairs)

# ←←← Time 図だけ上限に余白を追加（必ず add_sig_brackets の“後”に適用）
if HEADROOM_RATIO_TIME and HEADROOM_RATIO_TIME != 0:
    ymin, ymax = axes[0].get_ylim()
    axes[0].set_ylim(ymin, ymax + (ymax - ymin) * HEADROOM_RATIO_TIME)

# === 2️⃣ N_cmds ===
metric = "N_cmds"
stat, p, pairs, pivot = friedman_nemenyi(df, metric)
vals = [pivot[c].dropna().values for c in COND_ORDER]
bp = axes[1].boxplot(vals, labels=COND_ORDER, patch_artist=True, showfliers=False)
for i, patch in enumerate(bp['boxes']):
    patch.set_facecolor(palette[i]); patch.set_alpha(0.85)
    patch.set_edgecolor("black");    patch.set_linewidth(1.2)
for whisker in bp['whiskers']: whisker.set_color("black")
for cap in bp['caps']:         cap.set_color("black")
for median in bp['medians']:   median.set_color("black")
for i, v in enumerate(vals, 1):
    xs = np.random.normal(loc=i, scale=0.04, size=len(v))
    axes[1].plot(xs, v, 'o', color='black', alpha=0.75, ms=4)
axes[1].set_title(f"Number of Commands ($N_{{cmds}}$) — χ²={stat:.2f}, p={p:.3g}")
axes[1].set_ylabel("Count")
axes[1].grid(alpha=0.35, linestyle='--')
add_sig_brackets(axes[1], pairs)

# === 3️⃣ 散布図（色は条件ごとに Set2 を割当） ===
pivot_T = df.pivot(index="participant", columns="condition", values="T_task_sec").reindex(columns=COND_ORDER)
pivot_N = df.pivot(index="participant", columns="condition", values="N_cmds").reindex(columns=COND_ORDER)
x, y, c = [], [], []
for i, cond in enumerate(COND_ORDER):
    merged = pd.concat([pivot_T[cond].rename("T"), pivot_N[cond].rename("N")], axis=1).dropna()
    x.extend(merged["N"]); y.extend(merged["T"]); c.extend([palette[i]] * len(merged))
axes[2].scatter(x, y, color=c, edgecolor="black", alpha=0.8, s=40)
axes[2].set_xlabel("Number of Commands ($N_{cmds}$)")
axes[2].set_ylabel("Task Completion Time ($T_{task}$)")
axes[2].set_title("Relationship between $N_{cmds}$ and $T_{task}$")
axes[2].grid(alpha=0.35, linestyle='--')

plt.tight_layout()
plt.show()
```
