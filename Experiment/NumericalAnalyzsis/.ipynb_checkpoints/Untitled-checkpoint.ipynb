```python
import json
from pprint import pprint

# JSONファイルを読み込む
with open("./ArrangeData/PreTaskArrangementA.json", "r", encoding="utf-8") as f:
    data = json.load(f)


# 各arrangementごとに色の種類数を集計
for arrangement in data:
    arrange_id = arrangement["device_arrange_id"]
    arrange_name = arrangement.get("device_arrange_name", "")
    devices = arrangement.get("devices", [])

    # 各arrangement内で登場する色の種類を抽出
    colors = {d["colorName"] for d in devices}

    print(f"📘 {arrange_name} ({arrange_id})")
    print(f"  色の種類数: {len(colors)} 種類")
    print(f"  色の一覧: {sorted(colors)}\n")
```

```python
import json
import glob
import os

# ArrangeDataフォルダ内の全JSONファイルを対象
folder = "./ArrangeData"
pattern = os.path.join(folder, "PreTaskArrangement*.json")

# 結果を格納する辞書
arrange_color_map = {}

# 全ファイルを順に読み込み
for file_path in sorted(glob.glob(pattern)):
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    for arrangement in data:
        arrange_id = arrangement["device_arrange_id"]
        devices = arrangement.get("devices", [])
        colors = {d["colorName"] for d in devices}
        arrange_color_map[arrange_id] = sorted(colors)

# 出力確認
print("=== device_arrange_id → 色の種類 ===")
for aid, colors in arrange_color_map.items():
    print(f"{aid}: {colors}")

# 必要ならJSONとして保存
with open("arrange_color_map.json", "w", encoding="utf-8") as f:
    json.dump(arrange_color_map, f, ensure_ascii=False, indent=2)
```

```python
import os
import json
import pandas as pd
from pathlib import Path

# --- 設定 ---
RESULTS_ROOT = Path("./RESULTS")
OUT_CSV = Path("./imwut_results/system_time_summary.csv")
COND_MAP = {1: "SR", 2: "P+SR", 3: "Pointing", 4: "Label"}


def safe_get(d, *keys, default=None):
    cur = d
    for k in keys:
        if not isinstance(cur, dict):
            return default
        cur = cur.get(k)
    return cur if cur is not None else default


def load_participant(participant_id: int) -> pd.DataFrame:
    """ ./RESULTS/Pxx/Pxx_1.json ～ Pxx_4.json を読み取る """
    pdir = RESULTS_ROOT / f"P{participant_id}"
    if not pdir.exists():
        return pd.DataFrame()

    rows = []
    for cond_num, cond_label in COND_MAP.items():
        fpath = pdir / f"P{participant_id}_{cond_num}.json"
        if not fpath.exists():
            continue

        try:
            with open(fpath, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f"[WARN] failed to load {fpath}: {e}")
            continue

        # {results: [...]} 形式 or list 形式の両方に対応
        items = data["results"] if isinstance(data, dict) and "results" in data else data
        if not isinstance(items, list):
            continue

        for it in items:
            task_id = it.get("task_id") or it.get("taskId")
            sys_t = (
                safe_get(it, "metrics", "tokens", "system_total_time_sec")
                or safe_get(it, "metrics", "system_total_time_sec")
            )
            if sys_t is None:
                continue

            rows.append({
                "participant": participant_id,
                "condition": cond_label,
                "taskId": task_id,
                "system_time_total_sec": float(sys_t),
            })

    return pd.DataFrame(rows)


def main():
    all_rows = []

    for pid in range(1, 25):  # P1〜P24
        df = load_participant(pid)
        if df.empty:
            print(f"[WARN] No data for P{pid}")
            continue
        print(f"[OK] Loaded P{pid}: {len(df)} records")
        all_rows.append(df)

    if not all_rows:
        print("⚠️ No data found at all.")
        return

    df_all = pd.concat(all_rows, ignore_index=True)

    # タスク単位で合計時間を集計
    df_summary = (
        df_all.groupby(["participant", "condition", "taskId"], as_index=False)
              .agg(system_time_total_sec=("system_time_total_sec", "sum"),
                   n_attempts=("taskId", "count"))
    )

    os.makedirs(OUT_CSV.parent, exist_ok=True)
    df_summary.to_csv(OUT_CSV, index=False)
    print(f"[DONE] {OUT_CSV} written ({len(df_summary)} rows)")

    # 確認用
    print(df_summary.head())



main()
```
