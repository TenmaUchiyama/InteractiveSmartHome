```python
import json
from pprint import pprint

# JSONファイルを読み込む
with open("./ArrangeData/PreTaskArrangementA.json", "r", encoding="utf-8") as f:
    data = json.load(f)


# 各arrangementごとに色の種類数を集計
for arrangement in data:
    arrange_id = arrangement["device_arrange_id"]
    arrange_name = arrangement.get("device_arrange_name", "")
    devices = arrangement.get("devices", [])

    # 各arrangement内で登場する色の種類を抽出
    colors = {d["colorName"] for d in devices}

    print(f"📘 {arrange_name} ({arrange_id})")
    print(f"  色の種類数: {len(colors)} 種類")
    print(f"  色の一覧: {sorted(colors)}\n")
```

```python
import json
import glob
import os

# ArrangeDataフォルダ内の全JSONファイルを対象
folder = "./ArrangeData"
pattern = os.path.join(folder, "PreTaskArrangement*.json")

# 結果を格納する辞書
arrange_color_map = {}

# 全ファイルを順に読み込み
for file_path in sorted(glob.glob(pattern)):
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    for arrangement in data:
        arrange_id = arrangement["device_arrange_id"]
        devices = arrangement.get("devices", [])
        colors = {d["colorName"] for d in devices}
        arrange_color_map[arrange_id] = sorted(colors)

# 出力確認
print("=== device_arrange_id → 色の種類 ===")
for aid, colors in arrange_color_map.items():
    print(f"{aid}: {colors}")

# 必要ならJSONとして保存
with open("arrange_color_map.json", "w", encoding="utf-8") as f:
    json.dump(arrange_color_map, f, ensure_ascii=False, indent=2)
```

```python
import os
import json
import pandas as pd
from pathlib import Path

# ルートパス
ROOT = Path("./RESULTS")

# 空のリストを用意
records = []

# 全参加者を走査
for p_dir in sorted(ROOT.glob("P*")):
    pid = p_dir.name  # 例: "P21"
    for json_path in p_dir.glob(f"{pid}_*.json"):
        path_as_string = str(json_path)
    
       
        if not ("15" in path_as_string and "16" in path_as_string):
            continue
        try:
            # 条件番号をファイル名から取得 (例: P21_3.json → 3)
            cond_num = int(json_path.stem.split("_")[-1])

            # JSONをロード
            with open(json_path, "r", encoding="utf-8") as f:
                data = json.load(f)

            # 直接list or dict["results"]
            if isinstance(data, dict) and "results" in data:
                items = data["results"]
            else:
                items = data

            # 要素ごとに展開
            for i, item in enumerate(items):
                records.append({
                    "participant": pid,
                    "condition_num": cond_num,
                    "data_index": i,
                    "content": item
                })

            print(f"[OK] Loaded {json_path}")
        except Exception as e:
            print(f"[WARN] Failed to load {json_path}: {e}")

# pandas DataFrame 化
df = pd.DataFrame(records)
print("✅ DataFrame created:")
print(df.head())

# 条件番号でソート
df = df.sort_values(["participant", "condition_num"])

df
```

```python
import json
from pathlib import Path

# 任意のファイルを指定
path = Path("./RESULTS/P21/P21_1.json")

# ファイルが存在するかチェック
if not path.exists():
    print(f"❌ File not found: {path}")
else:
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    print("✅ JSON loaded successfully")
    print("type:", type(data))

    # 最初の要素を確認（リストの場合）
    if isinstance(data, list) and len(data) > 0:
        print("First element keys:", data[0].keys())
        print("Sample data:", json.dumps(data[0], indent=2, ensure_ascii=False)[:1000])
    elif isinstance(data, dict):
        print("Top-level keys:", data.keys())
        if "results" in data:
            results = data["results"]
            print(f"results length: {len(results)}")
            if results:
                print("First result keys:", results[0].keys())
                print("Sample result:", json.dumps(results[0], indent=2, ensure_ascii=False)[:1000])
    else:
        print("⚠️ Unknown structure:", type(data))
```
