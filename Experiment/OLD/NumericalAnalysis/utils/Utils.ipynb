# AttemptのUnique

```python
import os
import json
import shutil
from datetime import datetime

# タイムスタンプ付きバックアップディレクトリ名
timestamp = datetime.now().strftime("backup_%Y%m%d_%H%M%S")

# ベースディレクトリ
base_dir = r"C:\Users\Tenma\Desktop\Keio\InteractiveSmartHome\InteractiveSmartHome\InteractiveSmartHome\Assets\EXPERIMENT"
voice_log_dir = os.path.join(base_dir, "VOICE_LOG")
backup_root_dir = os.path.join(base_dir, "OLD", timestamp)

# バックアップルート作成
os.makedirs(backup_root_dir, exist_ok=True)

# 各Pディレクトリを処理
for participant in os.listdir(voice_log_dir):
    participant_path = os.path.join(voice_log_dir, participant)
    if not os.path.isdir(participant_path):
        continue

    # バックアップ先の参加者フォルダ作成
    backup_participant_path = os.path.join(backup_root_dir, participant)
    os.makedirs(backup_participant_path, exist_ok=True)

    for filename in os.listdir(participant_path):
        if not filename.endswith(".json"):
            continue

        file_path = os.path.join(participant_path, filename)
        backup_path = os.path.join(backup_participant_path, filename)

        # JSON読み込み
        try:
            with open(file_path, "r", encoding="utf-8") as f:
                data = json.load(f)
        except json.JSONDecodeError:
            print(f"❌ JSON decode error: {file_path}")
            continue

        modified = False

        # 各タスクごとに attemptId をユニークにする
        for task in data:
            seen_ids = set()
            unique_attempts = []

            for attempt in task.get("taskAttempts", []):
                aid = attempt.get("attemptId")
                if aid not in seen_ids:
                    seen_ids.add(aid)
                    unique_attempts.append(attempt)

            if len(unique_attempts) != len(task.get("taskAttempts", [])):
                modified = True
                task["taskAttempts"] = unique_attempts
                task["taskAttemptCount"] = len(unique_attempts)
                task["finalId"] = unique_attempts[-1]["attemptId"] if unique_attempts else ""

        if modified:
            # バックアップ
            shutil.copy2(file_path, backup_path)
            # 上書き保存
            with open(file_path, "w", encoding="utf-8") as f:
                json.dump(data, f, ensure_ascii=False, indent=2)
            print(f"✅ Cleaned & Saved: {file_path}")
        else:
            print(f"⏭️ No duplicate: {file_path}")
```

# VoiceLogの名前変更

```python
import os
import shutil
from datetime import datetime

# タイムスタンプ付きバックアップディレクトリ名
timestamp = datetime.now().strftime("backup_%Y%m%d_%H%M%S")

# パス設定
base_dir = r"C:\Users\Tenma\Desktop\Keio\InteractiveSmartHome\InteractiveSmartHome\InteractiveSmartHome\Assets\EXPERIMENT"
voice_log_dir = os.path.join(base_dir, "VOICE_LOG")
backup_root_dir = os.path.join(base_dir, "OLD", timestamp)

# バックアップディレクトリ作成
os.makedirs(backup_root_dir, exist_ok=True)

# ディレクトリ（P1〜P22）をループ
for participant in os.listdir(voice_log_dir):
    participant_path = os.path.join(voice_log_dir, participant)
    
    if not os.path.isdir(participant_path):
        continue

    # バックアップ先ディレクトリを作成
    backup_participant_path = os.path.join(backup_root_dir, participant)
    os.makedirs(backup_participant_path, exist_ok=True)

    for filename in os.listdir(participant_path):
        if filename.endswith(".json") and "_" in filename:
            num_part = filename.split("_")[0]
            new_filename = f"{participant}_{num_part}.json"

            old_path = os.path.join(participant_path, filename)
            new_path = os.path.join(participant_path, new_filename)
            backup_path = os.path.join(backup_participant_path, filename)

            # バックアップコピー
            shutil.copy2(old_path, backup_path)
            print(f"✅ Backed up: {filename} → {backup_path}")

            # すでに存在していたらスキップ
            if os.path.exists(new_path):
                print(f"⚠️ Skip (already exists): {new_filename}")
                continue

            # リネーム実行
            try:
                os.rename(old_path, new_path)
                print(f"✏️ Renamed: {filename} → {new_filename}")
            except Exception as e:
                print(f"❌ Error renaming {filename}: {e}")
```

```python
import os
import json
import shutil
from datetime import datetime

# タイムスタンプ付きバックアップフォルダ作成
timestamp = datetime.now().strftime("backup_%Y%m%d_%H%M%S")
base_dir = r"C:\Users\Tenma\Desktop\Keio\InteractiveSmartHome\InteractiveSmartHome\InteractiveSmartHome\Assets\EXPERIMENT"
voice_log_dir = os.path.join(base_dir, "VOICE_LOG")
backup_dir = os.path.join(base_dir, "OLD", timestamp)
os.makedirs(backup_dir, exist_ok=True)

participants = [f"P{i}" for i in range(1, 22)]
conditions = [1, 2, 3, 4]

for participant in participants:
    participant_dir = os.path.join(voice_log_dir, participant)
    if not os.path.isdir(participant_dir):
        continue

    backup_participant_dir = os.path.join(backup_dir, participant)
    os.makedirs(backup_participant_dir, exist_ok=True)

    for condition in conditions:
        file_name = f"{participant}_{condition}.json"
        file_path = os.path.join(participant_dir, file_name)
        backup_path = os.path.join(backup_participant_dir, file_name)

        if not os.path.exists(file_path):
            print(f"❌ ファイルが見つかりません: {file_path}")
            continue

        with open(file_path, "r", encoding="utf-8") as f:
            try:
                data = json.load(f)
            except json.JSONDecodeError:
                print(f"❌ JSON読み込みエラー: {file_path}")
                continue

        modified = False

        for task in data:
            original_attempts = task.get("taskAttempts", [])
            filtered_attempts = [
                attempt for attempt in original_attempts
                if attempt.get("outputDevices")  # 空リスト・Noneを除外
            ]

            if len(filtered_attempts) != len(original_attempts):
                modified = True
                task["taskAttempts"] = filtered_attempts
                task["taskAttemptCount"] = len(filtered_attempts)
                task["finalId"] = filtered_attempts[-1]["attemptId"] if filtered_attempts else ""

        if modified:
            shutil.copy2(file_path, backup_path)
            with open(file_path, "w", encoding="utf-8") as f:
                json.dump(data, f, ensure_ascii=False, indent=2)
            print(f"✅ 修正・保存済み: {file_path}")
        else:
            print(f"🆗 出力ありAttemptのみ、変更なし: {file_path}")
```
