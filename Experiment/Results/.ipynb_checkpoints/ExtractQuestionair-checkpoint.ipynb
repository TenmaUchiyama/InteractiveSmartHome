```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("SurvayResult.csv")

# --- 定数定義 ---
CATEGORY_COLUMNS = {
    'sus':     list(range(1, 11)),   # SUS 1–10
    'tam':     list(range(11, 14)),  # TAM 11–13
    'trust':   list(range(14, 18)),  # Trust 14–17
    'trust2':  list(range(18, 29)),  # Trust2 18–28
    'nasa':    list(range(29, 35)),  # NASA-TLX 29–37
}

print("--- 質問内容 ---")
for category, q_numbers in CATEGORY_COLUMNS.items():
    print(f"\n## カテゴリ: {category.upper()}")
    print("---")
    for q_num in q_numbers:
        # Assuming column names are 'Q1', 'Q2', etc.
        # If your actual column names are different, you'll need to adjust this.
        col_name = f'Q{q_num}'
        if col_name in df.columns:
            # Assuming the question content is in the first row of each question column.
            # If your header row directly contains the question, you might just need df.columns[df.columns.get_loc(col_name)]
            print(f"{col_name}: {df[col_name].iloc[0]}")
        else:
            print(f"警告: カラム '{col_name}' がデータフレームに見つかりません。")
```
