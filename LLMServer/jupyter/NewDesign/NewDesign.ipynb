```python
from dotenv import load_dotenv
import os

# .envファイルの絶対パスを指定
env_path = os.path.join( "..", "..", ".env")  # 親ディレクトリの.envを指定
load_dotenv(env_path)


```

```python
from langchain_openai import ChatOpenAI

filter_agent = ChatOpenAI(model="gpt-4o", verbose=True)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

with open("./filter_agent.txt", "r", encoding="utf-8") as f:
    data = SystemMessage(f.read())
```

```python
human_message = HumanMessage(content="後ろの右の電気つけて")
input_prompt= [
    data, 
    human_message
]
res = filter_agent.invoke(input_prompt)
```
