```python
import json
import paho.mqtt.client as mqtt
```

```python
MQTT_BROKER = "localhost"  # MQTTブローカーのホスト名
MQTT_PORT = 1883  # MQTTポート番号（通常は1883）
id = "0828dd3c-1046-4e24-b5bf-1ded8af1740f"
MQTT_TOPIC = "device/" + id  # 送信先のトピック

client = mqtt.Client()
client.connect(MQTT_BROKER, MQTT_PORT)
```

```python
def send(payload):
    """MQTTメッセージを送信"""
    try:
       
        client.publish(MQTT_TOPIC, json.dumps(payload))
        print(f"Message sent to MQTT broker '{MQTT_BROKER}' on topic '{MQTT_TOPIC}':")
        print(json.dumps(payload, indent=4))
    except Exception as e:
        print(f"Error sending MQTT message: {e}")
```

```python
light =  {
            "id": id,
            "state": True,
            "intensity": 100,
            "color": {"r": 255, "g": 0, "b": 0}  # Optional
        }


send(light)
```

```python
curtain =  {
            "id": id,
            "state": True,
            "intensity":0,
        }


send(curtain)
```
