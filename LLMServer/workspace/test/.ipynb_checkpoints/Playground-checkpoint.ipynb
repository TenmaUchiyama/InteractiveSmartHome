```python
from langchain_openai import ChatOpenAI 
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from typing import Annotated, List
from typing import Literal
from typing_extensions import TypedDict 
from langchain_core.tools import tool
class State(TypedDict):
    messages: Annotated[list, add_messages]
```

```python
llm = ChatOpenAI(model="gpt-4o-mini", verbose=True)
messages = [
    SystemMessage(content="あなたは優秀なアシスタントです。"),
    HumanMessage(content="こんにちは！わたしは承太郎です。"),
    AIMessage(content="こんにちは、承太郎さん！あなたのお役にたてるように頑張ります。なにか困ってることはありますか？"),
    HumanMessage(content="わたしの名前は？"),
]



ai_message = llm.invoke(messages)
ai_message
```

```python
from dataclasses import dataclass, asdict
from uuid import UUID, uuid4
from typing import Tuple
import json

@dataclass
class DeviceData:
    device_id: UUID
    device_name: str
    device_type: str
    mqtt_topic: str
    distance_from_user: float
    device_position: Tuple[float, float, float]


device = DeviceData(
    device_id=uuid4(),
    device_name="Living Room Light",
    device_type="Light",
    mqtt_topic="home/livingroom/light",
    distance_from_user = 3.9,
    device_position=(1.0, 2.0, 3.0)  # Vector3に相当
)

def device_data_to_json(device: DeviceData) -> str:
    # asdictを使ってdataclassを辞書に変換し、UUIDを文字列化する
    device_dict = asdict(device)
    device_dict['device_id'] = str(device_dict['device_id'])  # UUIDを文字列化
    return json.dumps(device_dict, indent=4)  # JSON文字列に変換



device_json = device_data_to_json(device)
print(device_json)
```
